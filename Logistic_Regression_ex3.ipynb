```python
# ============================================================
# AIM:
# To predict whether a customer will purchase a product
# using Logistic Regression.
# ============================================================

# -------------------- Import Libraries --------------------
import pandas as pd
from google.colab import files
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# -------------------- Upload Dataset --------------------
print("Upload the file: Social_Network_Ads.csv")
uploaded = files.upload()

# -------------------- Load Dataset --------------------
data = pd.read_csv("Social_Network_Ads.csv")

# -------------------- Display Dataset --------------------
print("\n========== FIRST 5 RECORDS ==========")
print(data.head())

print("\n========== DATASET INFORMATION ==========")
print(data.info())

print("\n========== MISSING VALUES ==========")
print(data.isnull().sum())

# -------------------- Select Features --------------------
# Independent Variables
X = data[['Age', 'EstimatedSalary']]

# Dependent Variable
y = data['Purchased']

# -------------------- Split Dataset --------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# -------------------- Train Model --------------------
model = LogisticRegression(max_iter=1000)

model.fit(X_train, y_train)

# -------------------- Predict Test Data --------------------
prediction = model.predict(X_test)

# -------------------- Evaluate Model --------------------
print("\n========== MODEL ACCURACY ==========")
print("Accuracy :", accuracy_score(y_test, prediction))

print("\n========== CONFUSION MATRIX ==========")
print(confusion_matrix(y_test, prediction))

print("\n========== CLASSIFICATION REPORT ==========")
print(classification_report(y_test, prediction))

# -------------------- User Input --------------------
print("\n========== ENTER CUSTOMER DETAILS ==========")

age = int(input("Enter Customer Age : "))
salary = float(input("Enter Estimated Salary : "))

customer = pd.DataFrame({
    "Age": [age],
    "EstimatedSalary": [salary]
})

# -------------------- Predict --------------------
result = model.predict(customer)
probability = model.predict_proba(customer)

print("\n========== PREDICTION ==========")
print("Probability of NOT Purchasing : {:.2f}%".format(probability[0][0] * 100))
print("Probability of Purchasing     : {:.2f}%".format(probability[0][1] * 100))

if result[0] == 1:
    print("\nPrediction : Customer is likely to PURCHASE the product.")
else:
    print("\nPrediction : Customer is NOT likely to purchase the product.")

# -------------------- Result --------------------
print("\n========== RESULT ==========")
print("Customer purchase prediction using Logistic Regression was implemented successfully.")
```
